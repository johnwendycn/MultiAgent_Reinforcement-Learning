## 4. Run Traceability Logs
Audits run metadata JSONs linking evaluation CSVs to model checkpoints.

```python
logs_dir = Path("../results/logs")
if not logs_dir.exists():
    logs_dir = Path("results/logs")

log_files = list(logs_dir.glob("*.json"))
print(f"Found {len(log_files)} run audit logs:")
for log_file in log_files:
    with open(log_file, "r", encoding="utf-8") as f:
        meta = json.load(f)
    print(f"  • Run ID: {meta.get('run_id')} | Created: {meta.get('created_at_utc')} | Status: {meta.get('status')}")
    print(f"    Checkpoints recorded: {len(meta.get('checkpoints', []))}")
    print(f"    Raw CSVs linked:      {len(meta.get('raw_eval_csvs', []))}")
```
